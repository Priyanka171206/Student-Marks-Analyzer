```python
# INSTALL REQUIRED LIBRARIES
%mamba install pandas

# IMPORT LIBRARIES
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# 1. STUDENT NAMES
students = [
    "Aman", "Neha", "Riya", "Simran", "Karan",
    "Rahul", "Priya", "Arjun", "Isha", "Vikas",
    "Meena", "Rohit", "Anjali", "Varun", "Kavya",
    "Nikhil", "Pooja", "Aditya", "Sneha", "Manish",
    "Tanya", "Deepak", "Sakshi", "Mohit", "Aarav",
    "Divya", "Harsh", "Naina", "Yash", "Komal"
]

# 2. SUBJECTS
subjects = [
    "Python", "Mathematics", "Data Structures",
    "Computer Networks", "DBMS"
]

# 3. STUDENT MARKS
marks = np.array([
    [78,85,72,80,88],
    [45,38,52,60,55],
    [88,91,84,86,90],
    [92,86,84,89,90],
    [60,55,67,72,65],
    [75,80,78,82,76],
    [89,92,95,88,91],
    [56,62,58,65,60],
    [81,79,85,88,82],
    [67,72,69,74,70],
    [90,88,92,94,89],
    [42,48,45,51,47],
    [76,82,79,85,80],
    [63,59,66,70,64],
    [85,87,90,84,88],
    [72,68,75,78,73],
    [95,93,91,96,94],
    [51,56,49,58,54],
    [83,86,80,89,85],
    [69,74,71,68,72],
    [87,90,85,92,88],
    [58,61,55,63,60],
    [79,84,82,80,86],
    [48,52,46,55,50],
    [91,89,94,90,92],
    [74,77,73,79,76],
    [65,70,68,72,67],
    [86,83,88,85,90],
    [53,57,60,55,58],
    [80,82,84,81,85]
])

# 4. TOTAL AND AVERAGE
total_marks = np.sum(marks, axis=1)
average_marks = np.mean(marks, axis=1)

# 5. GRADES AND RESULTS
grades = []
results = []

for i in range(len(students)):

    if np.any(marks[i] < 33):
        results.append("FAIL")
        grades.append("F")

    else:
        results.append("PASS")

        if average_marks[i] >= 90:
            grades.append("A+")
        elif average_marks[i] >= 80:
            grades.append("A")
        elif average_marks[i] >= 70:
            grades.append("B")
        elif average_marks[i] >= 60:
            grades.append("C")
        elif average_marks[i] >= 50:
            grades.append("D")
        else:
            grades.append("E")

# 6. TOPPER AND LOWEST SCORER
topper_index = np.argmax(total_marks)
lowest_index = np.argmin(total_marks)

print("========== CLASS PERFORMANCE ==========")
print("Class Topper:", students[topper_index])
print("Highest Marks:", total_marks[topper_index])

print("\nLowest Scorer:", students[lowest_index])
print("Lowest Marks:", total_marks[lowest_index])

# 7. SUBJECT PERFORMANCE
subject_average = np.mean(marks, axis=0)

best_subject = np.argmax(subject_average)
weak_subject = np.argmin(subject_average)

print("\nBest Subject:", subjects[best_subject])
print("Weakest Subject:", subjects[weak_subject])

# 8. PASS AND FAIL ANALYSIS
pass_count = results.count("PASS")
fail_count = results.count("FAIL")

print("\nTotal Students:", len(students))
print("Passed Students:", pass_count)
print("Failed Students:", fail_count)

# 9. COMPLETE RESULT TABLE
data = pd.DataFrame(marks, columns=subjects)

data.insert(0, "Student Name", students)
data["Total Marks"] = total_marks
data["Percentage"] = average_marks
data["Grade"] = grades
data["Result"] = results

print("\n========== STUDENT RESULT TABLE ==========")
display(data)

# 10. GRAPH 1: STUDENT PERFORMANCE
plt.figure(figsize=(14, 6))
plt.bar(students, total_marks)

plt.title("Student Performance Analysis")
plt.xlabel("Students")
plt.ylabel("Total Marks (Out of 500)")
plt.xticks(rotation=45)
plt.ylim(0, 500)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

# 11. GRAPH 2: SUBJECT PERFORMANCE
plt.figure(figsize=(10, 5))
plt.bar(subjects, subject_average)

plt.title("Subject-wise Class Performance")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.xticks(rotation=20)
plt.ylim(0, 100)
plt.grid(axis="y", linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()

# 12. GRAPH 3: PASS/FAIL ANALYSIS
plt.figure(figsize=(7, 7))

plt.pie(
    [pass_count, fail_count],
    labels=["Passed", "Failed"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Class Pass/Fail Analysis")
plt.show()

# 13. EXPORT CSV FILE
data.to_csv("student_results.csv", index=False)

print("\nStudent results exported successfully!")
```

```python
# Student Ranking
data["Rank"] = data["Total Marks"].rank(
    ascending=False, method="min"
).astype(int)

ranked_students = data.sort_values("Rank")

print("STUDENT RANKING")
display(ranked_students[
    ["Rank", "Student Name", "Total Marks", "Percentage", "Grade"]
])
```

```python
print("SUBJECT-WISE TOPPERS")

for subject in subjects:
    topper_index = np.argmax(data[subject])
    
    print(
        subject, ":",
        students[topper_index],
        "-", data[subject].max(), "marks"
    )
```

```python
grade_counts = data["Grade"].value_counts().sort_index()

plt.figure(figsize=(7, 5))
plt.pie(
    grade_counts,
    labels=grade_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Grade Distribution")
plt.show()
```

```python
plt.figure(figsize=(10, 8))

plt.imshow(marks, cmap="YlGnBu", aspect="auto")

plt.colorbar(label="Marks")
plt.xticks(range(len(subjects)), subjects, rotation=30)
plt.yticks(range(len(students)), students)

plt.title("Student Performance Heatmap")
plt.xlabel("Subjects")
plt.ylabel("Students")

plt.tight_layout()
plt.show()
```

```python
print("===== CLASS PERFORMANCE SUMMARY =====")

print("Class Average:", round(average_marks.mean(), 2))
print("Highest Marks:", total_marks.max())
print("Lowest Marks:", total_marks.min())

passed = results.count("PASS")
failed = results.count("FAIL")

print("Total Passed:", passed)
print("Total Failed:", failed)
print("Pass Percentage:", round((passed / len(students)) * 100, 2), "%")
```

```python
print("SUBJECT-WISE PASS/FAIL ANALYSIS")

for i, subject in enumerate(subjects):
    passed = np.sum(marks[:, i] >= 33)
    failed = len(students) - passed

    print("\n", subject)
    print("Passed:", passed)
    print("Failed:", failed)
```
